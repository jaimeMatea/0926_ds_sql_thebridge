```python
import pandas as pd
import sqlite3
```

```python
# conectamos con la base de datos my_database.db
connection = sqlite3.connect("database_cervezas.db")

# obtenemos un cursor que utilizamos para las queries
crsr = connection.cursor()
```

```python
# Con esta función leemos los datos y lo pasamos a un DataFrame de Pandas
def sql_query(query):

    # Ejecuta la query
    crsr.execute(query)

    # Almacena los datos de la query 
    ans = crsr.fetchall()

    # Obtenemos los nombres de las columnas de la tabla
    names = [description[0] for description in crsr.description]

    return pd.DataFrame(ans,columns=names)
```

```python
# 1 Obtener el nombre de los empleados que hayan repartido al bar Stop durante la semana del 17 al 23 de octubre de 2005.

query = '''
    select e.Nombre
    from empleados e
    join reparto r on e.CodE=r.CodE
    join bares b on b.CodB=r.CodB
    where b.Nombre='Stop' and r.Fecha between '10/17/05' and '10/23/05'
'''
sql_query(query)
```

```python
# 2 Obtener el Cif y nombre de los bares a los que se ha repartido cerveza de tipo Botella y 
# capacidad inferior a 1 litro, ordenados por localidad

query = '''
    select distinct b.Nombre,b.Cif
    from bares b
    join reparto r on b.CodB=r.CodB
    join cervezas c on r.CodC=c.CodC
    where c.Envase='Botella' and c.Capacidad<1
    order by b.Localidad asc;
'''
sql_query(query)
```

```python
# 3. Obtener los repartos (nombre del bar, envase y capacidad de la bebida, fecha y cantidad) realizados por Carlos Lopez
query = '''
    select b.Nombre, c.Envase, c.Capacidad, r.Fecha, r.Cantidad
    from reparto r
    join bares b on r.CodB=b.CodB
    join cervezas c on r.CodC=c.CodC
    join empleados e on r.CodE=e.CodE
    where e.Nombre="Carlos Lopez"
'''
sql_query(query)
```

```python
# 4. Obtener los bares a los que se les ha repartido envases de tipo botella y capacidad 0.2 o 0.33
query = '''
    select b.Nombre
    from bares b
    join reparto r on b.CodB=r.CodB
    join cervezas c on r.CodC=c.CodC
    where c.Envase="Botella" and (c.Capacidad=0.2 or c.Capacidad=0.33)
'''
sql_query(query)
```

```python
# 5. Nombre de los empleados que han repartido a los bares "Stop" y "Las Vegas" cervezas con envase botella.
query = '''
    select distinct b.Nombre as "Nombre bar" ,e.Nombre as "Nombre empleado", c.Envase 
    from empleados e
    join reparto r on e.CodE=r.CodE
    join bares b on r.CodB=b.Codb
    join cervezas c on r.CodC=c.CodC
    where b.Nombre in ("Stop","Las Vegas") and c.Envase="Botella"
'''
sql_query(query)
```

```python
# 6. Obtener el nombre y número de viajes que ha realizado cada empleado fuera de Villa Botijo.
query = '''
    select e.CodE, e.Nombre, count(b.Localidad) as Viajes
    from empleados e
    join reparto r on e.CodE=r.CodE
    join bares b on r.CodB=b.CodB
    where not Localidad="Villa Botijo"
    group by e.Nombre
'''
sql_query(query)
```

```python
# 7. Obtener el nombre y localidad del bar que más litros de cerveza ha comprado.
query = '''
    WITH TotalLitrosPorBar AS (
    SELECT b.Nombre, b.Localidad, SUM(r.Cantidad * c.Capacidad) AS litros
    FROM bares b
    JOIN reparto r ON b.CodB = r.CodB
    JOIN cervezas c ON c.CodC = r.CodC
    GROUP BY b.CodB, b.Nombre, b.Localidad
)
SELECT Nombre, Localidad, litros
FROM TotalLitrosPorBar
WHERE litros = (SELECT MAX(litros) FROM TotalLitrosPorBar);
'''
sql_query(query)
```
